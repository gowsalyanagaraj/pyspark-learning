# pyspark dataframes filtering operations

- Filtering operations
- &, |, ==
- ~

```python
from pyspark.sql import SparkSession
spark = SparkSession.builder.appName("FilteringData").getOrCreate()
```

```python
df = spark.read.csv("../Datasets/test3.csv", header=True, inferSchema=True)
df.show()
```

### Filter operations

#### Salary of the people less than or equal to 50000

```python
df.filter("Salary<=50000").show()
```

```python
df.filter("Salary<=50000").select("Name", "Age").show()
```

```python
df.filter(df['Salary'] <= 50000).show()
```

```python
df.filter((df['Salary'] <= 50000) & 
          (df['Salary']>=30000)).show()
```

```python
df.filter(~(df['Salary'] <= 50000)).show()
```
