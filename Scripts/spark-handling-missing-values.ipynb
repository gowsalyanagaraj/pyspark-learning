# Handling missing values

### Import packages and read the file

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName('HandleMissingValues').getOrCreate()
```

```python
df = spark.read.csv('../Datasets/test2.csv', header=True, inferSchema=True)
```

```python
df.show()
```

### Drop the Columns

```python
df.drop('Name').show()
```

```python
df.show()
```

```python
df.na.drop().show()
```

#### how
- all => all the values are null then dropped
- any => any value is null then dropped

```python
df.na.drop(how='any').show()
```

#### threshold
- thresh=2 => atleast 2 non null must be there else dropped

```python
df.na.drop(how='any', thresh=3).show()
```

#### subset

```python
df.na.drop(how='any', subset=['Age']).show()
```

### Filling the missing values

```python
df.na.fill('Missing values').na.fill(0).show()
```

```python
df.na.fill({'Name': 'Missing', 'Age': 0, 'Salary': 0, 'Experience': 0}).show()
```

```python
df.show()
```

### Filling null values with mean

```python
!pip install numpy
```

```python
from pyspark.ml.feature import Imputer
```

```python
imputer = Imputer(
    inputCols = ['Age', 'Experience', 'Salary'],
    outputCols =[ '{}_imputed'.format(c) for c in ['Age', 'Experience', 'Salary']]
).setStrategy('median')
```

### Add imputation cols to df

```python
imputer.fit(df).transform(df).show()
```
