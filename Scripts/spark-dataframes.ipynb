### Adding columns in dataframes

```python
df = df.withColumn('Experience after 2 years', df['Experience']+2)
```

```python
df.show()
```

### drop the columns

```python
df = df.drop('Experience after 2 years')
```

```python
df.show()
```

### rename the columns

```python
df = df.withColumnRenamed('Name', 'Employee Name')
```

```python
df.show()
```
