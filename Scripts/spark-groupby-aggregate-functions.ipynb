# Pyspark groupBy and Aggregate Functions

```python
from pyspark.sql import SparkSession
```

```python
spark = SparkSession.builder.appName("AggregateFunctions").getOrCreate()
```

```python
df = spark.read.csv("../Datasets/test4.csv", header=True, inferSchema=True)
```

```python
df.show()
```

```python
df.printSchema()
```

### GroupBy

```python
df.groupBy('Name').sum('Salary').show()
```

##### Grouped to find the maximum salary

```python
df.groupBy("Name").max("Salary").show()
```

```python
df.groupBy("Name").min("Salary").show()
```

```python
df.groupBy("Name").avg("Salary").show()
```

##### GroupBy Departments which gives maximum salary

```python
df.groupBy("Department").sum("Salary").show()
```

```python
df.groupBy("Department").mean("Salary").show()
```

```python
df.groupBy("Department").count().show()
```

```python
df.agg({"Salary": "sum"}).show()
```
