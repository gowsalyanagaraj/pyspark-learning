```python
from pyspark.sql.functions import *
from pyspark.sql.types import *

df = spark.read.format('csv').options(header='true', inferSchema='true').load('/Volumes/workspace/default/firstvolume/BigMart Sales.csv')
df.display()
```

# Data Writing

### CSV

```python
df.write.format('csv').mode('append').save('/Volumes/workspace/default/firstvolume/sales_data.csv')
```

# Data Writing modes

1. Append
2. Overwrite
3. Error
4. Ignore
| **Save Mode Name** | **Behavioral Meaning** | **What it does if data already exists** |
|---|---|---|
| **`error`** or **`errorifexists`** | **Default Mode** | Throws an `AnalysisException`. |
| **`append`** | **Add Data** | Adds the DataFrame contents to the existing files/table rows. |
| **`overwrite`** | **Replace Data** | Deletes the old data entirely and replaces it with the new DataFrame. |
| **`ignore`** | **Skip Operation** | Silently does nothing and leaves the existing data untouched. |

#### Append

```python
df.write.format('csv')\
        .mode('append')\
        .save('/Volumes/workspace/default/firstvolume/sales_data.csv')
```

```python
df.write.format('csv')\
        .mode('append')\
        .option('path','/Volumes/workspace/default/firstvolume/sales_data.csv')\
        .save()
```

#### overwrite

```python
df.write.format('csv')\
        .mode('overwrite')\
        .save('/Volumes/workspace/default/firstvolume/sales_data.csv')
```

#### Error mode

```python
df.write.format('csv')\
        .mode('error')\
        .option('path', '/Volumes/workspace/default/firstvolume/sales_data.csv')\
        .save()
```

#### ignore mode

# Managed Vs External Tables

1. Managed Table
- A managed table is a table where Databricks manages both:
- Metadata: Table name, schema, columns, properties, etc.
- Data: The actual data files stored in the underlying storage.
When you create a managed table, Databricks automatically chooses the storage location based on the catalog and schema configuration.
- DROP TABLE -> Removes metadata and schedules data deletion

2. External Table
- An external table is a table where Databricks manages the metadata, but you manage the underlying data files.
- Metadata: Managed by Databricks.
- Data: Stored in a location that you specify in cloud storage.
- Storage location: You explicitly provide the location.
- Data lifecycle: You are responsible for managing the underlying data files.
- DROP TABLE -> Removes metadata only
